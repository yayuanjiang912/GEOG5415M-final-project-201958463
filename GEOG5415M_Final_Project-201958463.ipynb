## Housing affordability inequality in England (2008–2024)

**Research focus:** How does housing affordability (price-to-earnings) vary across space and deprivation groups, and what are the implications for public good?

## Executive summary
This project examines housing affordability inequality across England using Local Authority District (LAD) data from 2008–2024. I combine the UK House Price Index (October prices), ASHE median annual earnings, and the English Indices of Multiple Deprivation (IMD 2019) to construct a price-to-earnings affordability indicator and compare patterns across deprivation quintiles. Results show clear temporal change and strong spatial unevenness in 2024, indicating that housing pressure is not evenly shared across places. These findings are relevant for local authorities and housing policy teams because they highlight where affordability pressures are concentrated and where monitoring and targeted interventions may be most needed.

## 1. Research question and public good motivation
**Research questions:**
1) How does the price-to-earnings ratio vary across deprivation quintiles in England over 2008–2024?  
2) What does the spatial distribution of affordability pressure look like in 2024?

**Why this matters (public good):**
Housing affordability affects access to stable accommodation and shapes broader inequalities (e.g., displacement pressure, labour-market access, and wealth accumulation). Place-based disparities are especially important for public good because many housing levers—planning, local housing strategies, monitoring, and local support—operate at or are reported using local authority geographies.

**Primary audiences:** local authorities (housing strategy and planning teams), policymakers monitoring inequality, and the general public/civil society organisations concerned with housing justice.

## 2. Scope note (England-only)
This project focuses on England because the Indices of Multiple Deprivation (IMD) 2019 is an England-specific deprivation framework. For conceptual consistency, I restrict the analytical sample and mapping to England LADs (LAD codes with prefixes E06/E07/E08/E09).

## 3. Data sources, licences, and ethics
### 3.1 Data inventory
I use the following open datasets:
- **UK House Price Index (UK HPI)**: LAD-level October average price (P_oct) and derived annual log-change fields available in the provided panel.
- **ASHE (Annual Survey of Hours and Earnings)**: LAD-level median annual gross earnings (residence-based, full-time).
- **English Indices of Multiple Deprivation (IMD) 2019**: LAD-level deprivation summary and deprivation quintile grouping.
- **ONS LAD boundaries (May 2024, BFC)**: administrative boundary polygons used to create the 2024 map.

### 3.2 Ethics, bias, and representativeness
All datasets are aggregated to administrative areas and contain no personal identifiers, so privacy risks are low. However, several biases and limitations remain:
- **Measurement bias:** affordability depends on the choice of price measure (October price) and earnings definition (median annual gross, full-time, residence-based), which may not capture all households.
- **Aggregation bias / MAUP:** LAD-level statistics can mask within-area inequalities and neighbourhood-level affordability issues.
- **Time-invariance of deprivation:** IMD 2019 is used as a fixed proxy of deprivation for the full 2008–2024 period; relative deprivation may change over time.
- **Conceptual caution:** observed differences across deprivation groups are descriptive and do not imply causal mechanisms.

## 4. Methods overview
### 4.1 Data preparation and merging
I harmonise identifiers using the LAD `area_code`. I merge UK HPI with ASHE earnings by (area_code, year), then merge the resulting table with IMD 2019 by area_code, and compute the affordability indicator:
\[
\text{price-to-earnings} = \frac{P_{oct}}{\text{median annual earnings}}
\]

### 4.2 Visual analytics choices
To compare inequality across deprivation groups, I summarise affordability using the **median** price-to-earnings within each IMD quintile per year (robust to outliers). For spatial communication, I map the 2024 price-to-earnings ratio at LAD level to support place-based interpretation for policy audiences.

## 5. Data quality checks
I check missingness in the key variables (P_oct, earnings, IMD quintiles) and validate the boundary merge for mapping. The 2024 join to the LAD boundary achieves a high match rate (≈98.7%), indicating that only a small share of records are unmatched—likely due to boundary-code version differences or missing values.





## 6. Results
 ### 6.1 Affordability inequality over time (2008–2024)
Figure 1 shows that price-to-earnings ratios change substantially over time and differ across deprivation quintiles. This indicates that affordability pressure is not uniform: some deprivation groups consistently face higher house prices relative to local earnings. The time-series view also helps identify periods where affordability shifts coincide with broader macroeconomic phases (e.g., post-2008 adjustment and later tightening periods), although this analysis remains descriptive. Notably, the least deprived quintile (Q1) shows the highest price-to-earnings ratios across most years, consistent with higher house prices in more affluent areas relative to local earnings, while the most deprived quintile (Q5) tends to have lower ratios.

## 7. Discussion, limitations, and next steps
**Implications for public good:** The combined evidence of deprivation-group differences and 2024 spatial clustering suggests that affordability pressures are unevenly distributed and may reinforce wider spatial inequality. For local authorities, this supports prioritising affordability monitoring and aligning housing strategies with local earnings contexts rather than relying on prices alone.

**Limitations:** This project is descriptive and does not identify causal drivers. Affordability measurement is sensitive to indicator choices (October prices; full-time median earnings), and LAD aggregation can mask neighbourhood-level problems (MAUP). Small boundary-code mismatches also lead to a small share of unmatched records in the 2024 map (≈1.3%).

**Next steps:** With more time and data, the analysis could be extended by (i) using finer spatial units (e.g., LSOA), (ii) incorporating additional constraints such as housing supply indicators or rental affordability measures, and (iii) conducting formal modelling or quasi-experimental designs to test mechanisms rather than describing association.

```python
import pandas as pd
from pathlib import Path

DATA = Path("/content/project/data/raw")

HPI_PATH  = DATA / "uk_hpi_oct_panel_y_next.csv"
EARN_PATH = DATA / "earnings_ashe_residence_fulltime_median_annual_gross_england_lad_2005_2024_clean.csv"
IMD_PATH  = DATA / "imd2019_lad_clean.csv"

hpi  = pd.read_csv(HPI_PATH)
earn = pd.read_csv(EARN_PATH)
imd  = pd.read_csv(IMD_PATH)

print("hpi:", hpi.shape)
print("earn:", earn.shape)
print("imd:", imd.shape)
```

```python
for _df in [hpi, earn, imd]:
    _df["area_code"] = _df["area_code"].astype(str)

hpi["year"]  = pd.to_numeric(hpi["year"], errors="coerce")
earn["year"] = pd.to_numeric(earn["year"], errors="coerce")

eng_prefix = ("E06","E07","E08","E09")
hpi  = hpi[hpi["area_code"].str.startswith(eng_prefix)].copy()
earn = earn[earn["area_code"].str.startswith(eng_prefix)].copy()
imd  = imd[imd["area_code"].str.startswith(eng_prefix)].copy()

df = hpi.merge(
    earn[["area_code","year","earnings_median_annual_gross"]],
    on=["area_code","year"],
    how="left"
).merge(
    imd[["area_code","imd_avg_score","imd_quintile_score"]],
    on="area_code",
    how="left"
)

df["P_oct"] = pd.to_numeric(df["P_oct"], errors="coerce")
df["earnings_median_annual_gross"] = pd.to_numeric(df["earnings_median_annual_gross"], errors="coerce")
df["price_to_earnings"] = df["P_oct"] / df["earnings_median_annual_gross"]

df = df[df["year"].between(2008, 2024)].copy()
df["imd_quintile_score"] = df["imd_quintile_score"].astype(str).str.strip()

keep_cols = ["area_code","area_name","year","P_oct","earnings_median_annual_gross","imd_avg_score","imd_quintile_score","price_to_earnings"]
df = df[keep_cols].copy()

print("df shape (2008–2024):", df.shape)
print("Missing earnings (2008–2024):", df["earnings_median_annual_gross"].isna().mean())
print(df.head(3).to_string(index=False))
```

```python
import matplotlib.pyplot as plt
from pathlib import Path
import pandas as pd

OUT_FIG = Path("/content/project/outputs/figures")
OUT_FIG.mkdir(parents=True, exist_ok=True)

df_plot = df.dropna(subset=["price_to_earnings", "imd_quintile_score"]).copy()
df_plot["imd_quintile_score"] = df_plot["imd_quintile_score"].astype(str).str.strip()

order = ["Q1 least deprived", "Q2", "Q3", "Q4", "Q5 most deprived"]
df_plot["imd_quintile_score"] = pd.Categorical(df_plot["imd_quintile_score"], categories=order, ordered=True)

g = (df_plot
     .groupby(["year", "imd_quintile_score"], observed=True)["price_to_earnings"]
     .median()
     .reset_index())

plt.figure(figsize=(10, 6))
for q in order:
    sub = g[g["imd_quintile_score"] == q]
    if sub.empty:
        print(f"Warning: no data for {q}")
        continue
    plt.plot(sub["year"], sub["price_to_earnings"], label=q)

plt.title("Housing affordability (Price-to-earnings) by deprivation quintile, 2008–2024")
plt.xlabel("Year")
plt.ylabel("Price-to-earnings (P_oct / earnings)")
plt.legend()
plt.tight_layout()

fig1 = OUT_FIG / "fig_nonspatial_trend_imd_quintiles_2008_2024.png"
plt.savefig(fig1, dpi=300)
fig1
```
